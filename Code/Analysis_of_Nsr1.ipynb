```python
Df1=pd.read_excel("NSR1-GFP_size.xlsx")
plt.figure(figsize=(5,3))
ax=sns.violinplot(y='AreaShape_Area',x='plasmid',data=Df1, palette=["darkgrey", "lightseagreen"])

plt.xlabel(None)
plt.ylabel("Nsr1-GFP_size")
plt.ylim(-50, 1000)
plt.gcf().set_size_inches(3,3)

plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['bottom'].set_visible(True)
plt.gca().spines['left'].set_visible(True)

plt.show()
```

```python
Df1=pd.read_excel("20240805_NSR1_integratedintensity_perCell.xlsx")

plt.figure(figsize=(6, 8))
ax = sns.violinplot(x='plasmid', y='Intensity_IntegratedIntensity_NSR1', data=Df1, palette=["grey", "chocolate", "chocolate"])

plt.xlabel(None)
plt.ylabel("Nsr1-mScarlet-I fluorescence", fontsize=20)
plt.ylim(-20, 180)

plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['bottom'].set_visible(True)
plt.gca().spines['left'].set_visible(True)

plt.tick_params(axis='both', labelsize=20)

categories = Df1['plasmid'].unique()
for i, category in enumerate(categories):
    n = len(Df1[Df1['plasmid'] == category])
    median_value = Df1[Df1['plasmid'] == category]['Intensity_IntegratedIntensity_NSR1'].median()
    y_max = Df1['Intensity_IntegratedIntensity_NSR1'].max() + 5
    ax.text(i+0.4, median_value + 50, f'n={n}', horizontalalignment='center', verticalalignment='center', color='black', fontsize=15)

for i, violin in enumerate(ax.collections):
    if i == 2:
        violin.set_hatch('//')
        violin.set_edgecolor('white')
        violin.set_linewidth(1.5)

plt.show()
```

```python
Df1 = pd.read_excel("20240805_NSR1_size.xlsx")

ax = sns.violinplot(x='plasmid', y='AreaShape_Area', data=Df1, palette=["grey", "chocolate", "chocolate"])

plt.xlabel(None)
plt.ylabel("Nsr1-mScarlet-I size", fontsize=20)
plt.ylim(-100, 1000)
# plt.xticks(rotation=45)
plt.gcf().set_size_inches(6, 8)

plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['bottom'].set_visible(True)
plt.gca().spines['left'].set_visible(True)

# 目盛りラベルのフォントサイズを設定
plt.tick_params(axis='both', labelsize=20)

# 各バイオリンの左上にn数を表示
categories = Df1['plasmid'].unique()
for i, category in enumerate(categories):
    n = len(Df1[Df1['plasmid'] == category])
    median_value = Df1[Df1['plasmid'] == category]['AreaShape_Area'].median()
    y_max = Df1['AreaShape_Area'].max() + 5
    ax.text(i + 0.6, median_value + 300, f'n={n}',
            horizontalalignment='right',
            verticalalignment='center',
            color='black',
            fontsize=15)

# Gpm1_mut（Index: 2）にのみハッチングを追加
for i, violin in enumerate(ax.collections):
    if i == 2:
        violin.set_hatch('//')
        violin.set_edgecolor('white')
        violin.set_linewidth(1.5)

plt.show()
```
