```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("20240805_MajorMinor_ratio_cell.xlsx")

vector_data = df[df['plasmid'] == 'Vector']['MajorMinor_ratio']
gpm1_data = df[df['plasmid'] == 'Gpm1']['MajorMinor_ratio']
gpm1mut_data = df[df['plasmid'] == 'Gpm1mut']['MajorMinor_ratio']

plt.figure(figsize=(10, 3))
sns.kdeplot(vector_data, label='Vector', color='darkgray', fill=True, alpha=0.5)
sns.kdeplot(gpm1_data, label='Gpm1', color='brown', fill=True, alpha=0.5)
sns.kdeplot(gpm1mut_data, label='Gpm1-CCmut', color='orange', fill=True, alpha=0.5)

plt.xlabel("Axis ratio", fontsize=16)
plt.ylabel("Density", fontsize=16)
plt.xlim(0.6, 3)
plt.ylim(0, 5)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.legend(title="Plasmid", fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# データ読み込み
df = pd.read_excel("20240805_MajorMinor_ratio_cell.xlsx")
moxYG_data = df[df['plasmid'] == 'mox-YG']['MajorMinor_ratio']

# 描画の準備
plt.figure(figsize=(10, 3))

# 各グループの密度プロットを重ねてプロット
sns.kdeplot(vector_data, label='Vector', color='darkgray', fill=True, alpha=0.5)
sns.kdeplot(moxYG_data, label='mox-YG', color='seagreen', fill=True, alpha=0.5)

plt.xlabel("Major_length/Minor_length", fontsize=16)
plt.xlim(0.6, 3)
plt.ylim(0, 5)
plt.ylabel("Density", fontsize=16)
plt.legend(title="Plasmid", fontsize=12)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, alpha=0.3)

# グラフを保存
plt.savefig("20240805_MajorMinor_ratio_compared_GFP.pdf", bbox_inches='tight')
```
