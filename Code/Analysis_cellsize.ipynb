```python
Df1 = pd.read_excel("20240805_cellsize_from_RIP1.xlsx")

plt.figure(figsize=(6, 8))
ax = sns.violinplot(x='plasmid', y='AreaShape_Area', data=Df1, palette=["grey", "chocolate", "chocolate"])

plt.xlabel(None)
plt.ylabel("Cell size (AU)", fontsize=20)
plt.ylim(-1000, 14000)

plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['bottom'].set_visible(True)
plt.gca().spines['left'].set_visible(True)

plt.tick_params(axis='both', labelsize=20)

# n number in upper left corner of each violin
categories = Df1['plasmid'].unique()
for i, category in enumerate(categories):
    n = len(Df1[Df1['plasmid'] == category])
    median_value = Df1[Df1['plasmid'] == category]['AreaShape_Area'].median()
    y_max = Df1['AreaShape_Area'].max() + 5
    ax.text(i+0.4, median_value + 3500, f'n = {n}', horizontalalignment='center', verticalalignment='center', color='black', fontsize=15)

# Add hatching to Gpm1_mut
for i, violin in enumerate(ax.collections):
    if i == 2:
        violin.set_hatch('//')
        violin.set_edgecolor('white')
        violin.set_linewidth(1.5)

plt.show()
```
